```python
"""
Case study of Simple Banking System
Objective:
•	Practice Object-Oriented Programming (OOP) basics.
Instructions:
Study the following case: Simple Banking System. 
o	Write a small program using classes and objects.
o	Create three basic classes:
	Account (stores account number and balance).
	Customer (stores customer name and account).
	Transaction (stores transaction details like amount and type).
o	Add simple functions like:
	deposit(amount) – Adds money to the account.
	withdraw(amount) – Removes money if enough balance is available.
	display_balance() – Shows the account balance.
"""


#Define the Class
class Account:
    def __init__(self, account_number, balance=0):
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited Amount: ${amount}")
        print(f"New balance: ${self.balance}")

    def withdraw(self, amount):    
        if amount > self.balance:
            print("insufficient balance")
        else:
            self.balance -= amount
            print(f"Withdrawn Amount: ${amount}")
            print(f"New balance: ${self.balance}")

    def display_balance(self):
        print(f"Account {self.account_number}")
        print(f"Balance:${self.balance}")

class Customer:
    def __init__(self, name, account):
        self.name = name
        self.account = account

    def display_customer_info(self):
        print(f"Customer Name: {self.name}")
        self.account.display_balance()

class Transaction:
    def __init__(self, account, amount, transaction_type):
        self.account = account
        self.amount = amount
        self.transaction_type = transaction_type
        self.process_transaction()

    def process_transaction(self):
        if self.transaction_type == "deposit":
            self.account.deposit(self.amount)
        elif self.transaction_type == "withdraw":
            self.account.withdraw(self.amount)
        else:
            print("Invalid transaction type!")

#Test the functionality

# Creating an account for a customer
account1 = Account(account_number=101, balance=100)
customer1 = Customer(name="Alice", account=account1)

# Display customer details
customer1.display_customer_info()

# Performing transactions
transaction1 = Transaction(account1, 50, "deposit")
transaction2 = Transaction(account1, 30, "withdraw")

# Final account balance

print("\n Updated Account balance:")
customer1.display_customer_info()
```
