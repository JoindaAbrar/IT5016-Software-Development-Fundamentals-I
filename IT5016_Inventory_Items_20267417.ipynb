```python
#author: Joinda
#Subject: Software development (IT5016)
#Creating a Class for Inventory Items.


class InventoryItem:
    # -----------------------------
    # CLASS VARIABLES
    # -----------------------------
    # These belong to the class itself, not to individual objects.
    
    item_counter = 999          # Tracks the last used Item ID
    inventory = {}            # Stores all items using Item ID as the key


    # -----------------------------
    # 1: Add Inventory Item
    # -----------------------------
    def add_inventory_item(self):
        """
        This function:
        1. Generates a new Item ID
        2. Asks the user for item details
        3. Stores the item in the inventory dictionary
        4. Returns all item details
        """

        # STEP 1: Generate a unique Item ID
       
        InventoryItem.item_counter += 1   # By adding 1 each time (so IDs become 1000, 1001, 1002...)
        item_id = InventoryItem.item_counter

        # STEP 2: Collect item details from the user
        item_name = input("Enter Item Name: ")
        quantity = int(input("Enter Quantity: "))  # Convert to integer
        price_per_item = float(input("Enter Price per Item: $"))  # Convert to float

        # STEP 3: Store the item in the inventory dictionary
       
        InventoryItem.inventory[item_id] = {"name": item_name, "quantity": quantity, "price": price_per_item}

        # STEP 4: Return the collected values
        return item_name, quantity, price_per_item, item_id


    # -----------------------------
    # 2: Calculate Total Value
    # -----------------------------
    def calculate_total_value(self):
        """
        This function:
        1. Calls add_inventory_item()
        2. Calculates total value (quantity × price)
        3. Prints a summary
        4. Returns the total value
        """

        # STEP 1: Add item and receive its details
        item_name, quantity, price_per_item, item_id = self.add_inventory_item()

        # STEP 2: Calculate total value
        total_value = quantity * price_per_item

        # STEP 3: Display the result
        print("\nTotal Value Calculation:")
        print(f"Item Name: {item_name}")
        print(f"Item ID: {item_id}")
        print(f"Quantity: {quantity}")
        print(f"Price per Item: ${price_per_item}")
        print(f"Total Value: ${total_value:.2f}")

        return total_value


    # -----------------------------
    # 3: Update Inventory Item
    # -----------------------------
    def update_inventory(self, item_id):
        """
        This function:
        1. Checks if the item exists
        2. Allows user to update quantity and price
        3. Saves the updated values
        """

        # STEP 1: Check if item exists
        if item_id in InventoryItem.inventory:
            print(f"\nUpdating Item ID: {item_id}")

            # STEP 2: Ask for new values
            new_quantity = int(input("Enter new Quantity: "))
            new_price = float(input("Enter new Price per Item: $"))

            # STEP 3: Update the dictionary
            InventoryItem.inventory[item_id]["quantity"] = new_quantity
            InventoryItem.inventory[item_id]["price"] = new_price

            print("\nItem updated successfully.")
            return InventoryItem.inventory[item_id]

        else:
            print("Item not found.")
            return None


    # -----------------------------
    # 4: Display Inventory Item
    # -----------------------------
    def display_inventory_item(self, item_id):
        """
        This function:
        1. Checks if the item exists
        2. Displays all details including total value
        """

        # STEP 1: Check if item exists
        if item_id in InventoryItem.inventory:
            item = InventoryItem.inventory[item_id]

            # STEP 2: Calculate total value
            total_value = item["quantity"] * item["price"]

            # STEP 3: Display item details
            print("\nInventory Item Details")
            print("------------------------")
            print(f"Item Name   : {item['name']}")
            print(f"Item ID     : {item_id}")
            print(f"Quantity    : {item['quantity']}")
            print(f"Price/Item  : ${item['price']:.2f}")
            print(f"Total Value : ${total_value:.2f}")

        else:
            print("Item not found.")


# -----------------------------
# USING THE CLASS
# -----------------------------

# Create an object of InventoryItem
system = InventoryItem()

#Add item and calculate total value
system.calculate_total_value()

#Update item with ID 1000
system.update_inventory(1000)

#Display updated item
system.display_inventory_item(1000)
```
